# qiskit code for superposition

```python
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator
from qiskit.visualization import plot_histogram
circuit=QuantumCircuit(1,1)
circuit.h(0)
circuit.measure(0,0)
stimulator=AerSimulator()
result=stimulator.run(circuit,shots=1000).result()
counts=result.get_counts()
print("Measurement counts:",counts)
qc.draw("mpl")
#we use simulator to represent the output classically
```

# multiple qubit gates

```python
from qiskit import QuantumCircuit
qc=QuantumCircuit(3)
qc.h(0)
qc.cx(0,1)
qc.cz(1,2)
qc.swap(0,1)
print(qc.draw())
qc.draw("mpl")
```

# controlled -controlled gate

```python
from qiskit import QuantumCircuit
qc=QuantumCircuit(3)
qc.ccx(0,1,2) #toffoli gate
print(qc)
qc.draw("mpl")
```

# Rotation Gates

```python
from qiskit import QuantumCircuit
import numpy as pi
import numpy as np
qc=QuantumCircuit(1)
qc.rx(np.pi/2,0)
qc.ry(np.pi/2,0)
qc.rz(np.pi/2,0)
print(qc)
qc.draw("mpl")
```

# Measuring gates

```python
from qiskit import QuantumCircuit
qc=QuantumCircuit(2,2)
qc.h(0)
qc.cx(0,1)
qc.measure(0,0)
qc.measure(1,1)
print(qc)
qc.draw("mpl")
```

# GHZ state 3-qubit entanglement circuit

```python
from qiskit import QuantumCircuit
qc=QuantumCircuit(3,3)
qc.h(0)
qc.cx(0,1)
qc.cx(0,2)
print(qc)
#with measurement
qc.measure(0,0)
qc.measure(1,1)
qc.measure(2,2)
print(qc)
qc.draw("mpl")
#while measuring we must give (3,3) two values qubit and classical bit.
```
